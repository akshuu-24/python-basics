Operators

```python
#arithmetic operator
a = 15
b = 4

print("Addition:", a + b)  

print("Subtraction:", a - b) 

print("Multiplication:", a * b)  

print("Division:", a / b) 

print("Floor Division:", a // b)  

print("Modulus:", a % b) 

print("Exponentiation:", a ** b)
```

```python
#comparison operator
a = 13
b = 33

print(a > b) #greater than
print(a < b) #less than
print(a == b) #equal to
print(a != b) #not equal to
print(a >= b) # greater than and equal to
print(a <= b) #less than aequal to
```

```python
#logical operators
a = True
b = False
print(a and b)
print(a or b)
print(not a)
```

```python
#bitwise operator
a = 10
b = 4

print(a & b)#and
print(a | b)#or
print(~a) #not
print(a ^ b) #xor
```

```python
#assignment operator
a = 10
b = a
print(b)
b += a
print(b)
b -= a
print(b)
b *= a
print(b)
b <<= a
print(b)
```

```python
#identity operator
a = 10
b = 20
c = a

print(a is not b)
print(a is c)
```

```python
#membership operator
x = 24
y = 20
my_list = [10, 20, 30, 40, 50]

if (x not in my_list):
    print("x is NOT present in given list")
else:
    print("x is present in given list")

if (y in my_list):
    print("y is present in given list")
else:
    print("y is NOT present in given list")
```
