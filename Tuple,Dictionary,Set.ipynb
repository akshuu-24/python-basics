Tuple

```python
#Common Tuple Operations
#len(t)
t = (14,25,36,47,58,69,71,82,93)
print(len(t))
```

```python
#t.count(x)
t = (14,25,36,47,58,69,71,82,93)
print(t.count(47))
```

```python
#t.index(x)
t = (14,25,36,47,58,69,71,82,93)
print(t.index(71))
print(t.index(58))
```

```python
#t1+t2
t1 = (14,25,36,47)
t2 = (42,58,69,71,82,93)
result = t1+t2
print(result)
```

```python
#repetition
t1 = (14,25,36,47,58,69,71,82,93)
print(t1*2)
```

```python
#membership test
t1 = (14,25,36,47,58,69,71,82,93)
print(25 in t1)
```

DICTIONARY

#COMMON DICTIONARY OPERATIONS


```python
#D.KEYS()
S = {
    "A" : 1,
    "B" : 2
}
S.keys()
```

```python
#D.VALUES()
S = {
    "A" : 1,
    "B" : 2
}
S.values()
```

```python
#D.ITEMS()
S = {
    "A" : 1,
    "B" : 2
}
S.items()
```

```python
#D.pop()
S = {
    "A" : 1,
    "B" : 2
}
S.pop('A')
```

```python
#D.UPDATE()
S = {
    "A" : 1,
    "B" : 2
}
D2 = {'C':3, 'B':45}
S.update(D2)
print(S)
```

```python
#K in d
S = {
    "A" : 1,
    "B" : 2
}
'A' in S
```

SET

```python
#union
a = {1,2,3} 
b = {4,5,6}
union = a|b
print(union)
```

```python
#intersection
a = {1,2,3} 
b = {3,4,5,6}
intersection = a&b
print(intersection)
```

```python
#difference
a = {1,2,3} 
b = {3,4,5,6}
difference = a-b
print(difference)
```

```python
#symmetric diff.
a = {1,2,3} 
b = {3,4,5,6}
symdiff = a^b
print(symdiff)
```
