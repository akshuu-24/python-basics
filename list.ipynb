Indexing and slicing

```python
clrs = ["Orange","Blue","Buragndy","Cream","White","green"]
a =clrs[0]
print(a)
b = clrs[-1]
print(b)
c = clrs[1:3]
print(c)
d = clrs[:2]
print(d)
e = clrs[2:]
print(e)
f = clrs[::-1]
print(f)
```

```python
#common list methods
#APPEND
a = [1,2,3,4]
a.append(5)
print(a)
```

```python
#INSERT
q = [1,2,3,4,5]
q.insert(1,3)
print(q)
```

```python
#EXTEND
L = [1,2,3,4,5]
L.extend([6,7])
print(L)
```

```python
#POP
p = [1,2,3,4,5]
p.pop(1)
print(p)
```

```python
#SORT
K=[2,1,4,5,3]
K.sort()
print(K)
```

```python
#Reverse
M = [1,2,3,4,5]
M.reverse()
print(M)
```

```python
#clear
n = [1,2,3,4,5]
n.clear()
print(n)
```

```python
#list comprehensions
squares = []
for x in range(10):
    if x%2 == 0:
        squares.append(x*x)
        print(squares)
```
