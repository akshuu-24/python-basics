Tuple

```python

# Taking input
a = input("Enter elements separated by spaces: ").split()

# Converting list to tuple
t = tuple(a)
print(t)
```

```python
# Taking multiple integer inputs from the user
t = tuple(map(int, input("Enter elements separated by spaces: ").split()))
print(t)
```

```python
# Taking input
n = int(input("Enter the number of elements: "))
t = tuple(input(f"Enter element : ") for i in range(n))

print(t)
```

#Dictionary

```python
n = int(input("Enter the number of entries: "))
d = {input("Enter key: "): input("Enter value: ") for _ in range(n)}

print(d)
```

```python
n = int(input("Enter the number of entries: "))
entries = [(input("Enter key: "), input("Enter value: ")) for _ in range(n)]
d = dict(entries)

print(d)
```

```python
d = {} 
n = int(input("Enter the number of entries: "))

for _ in range(n):
    key = input("Enter key: ")
    value = input("Enter value: ")
    d.update({key: value})

print(d)
```
