```python
a=int(input("enter the number"))
for i in range(1,11):
    print(a * i)
```

```python
pr = [12.50,3.00,45.99,8.25]
total = 0
for p in pr: 
    total += p
    print(total)
```

```python
a = {"rahul","priya","amit"}
b = {"amit","sneha","rahul"}
cf = a & b
print("mutual_friends:")
for name in cf:
    print(f"friend: {name}")
```

While loop

```python
a=1
while a<10:
    print(a)
    a+=1
```

```python
name = "Akshat"
i = 1
while i <5:
    print(name)
    i += 1
```

```python
a = set()
count = 0
while count < 4:
    lang = input("enter language name:")
    a.add(lang)
    count += 1
    
print(f"total language:{len(a)}")
for lang in a:
    print(f"- {lang.upper()}")
```
